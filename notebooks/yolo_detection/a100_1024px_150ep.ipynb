```python
# Delete ALL macOS metadata files
import os
from pathlib import Path

total_removed = 0
for split in ["train", "val", "test"]:
    for subdir in ["images", "labels"]:
        d = Path(f"/workspace/dataset/{subdir}/{split}")
        if not d.exists():
            continue
        for f in d.glob("._*"):
            f.unlink()
            total_removed += 1

print(f"Removed {total_removed} macOS metadata files")

# Also check for .DS_Store files
for f in Path("/workspace/dataset").rglob(".DS_Store"):
    f.unlink()
    total_removed += 1

print(f"Total cleanup: {total_removed} files")
```

```python
import cv2
import matplotlib.pyplot as plt
from pathlib import Path


OUTPUT_DIR = Path("verification")   # relative path
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

candidates = ["721", "65", "982", "152", "437"]

for stem in candidates:
    sample_img = f"/workspace/dataset/images/train/{stem}.JPG"
    sample_lbl = f"/workspace/dataset/labels/train/{stem}.txt"
    
    img = cv2.imread(sample_img)
    if img is None:
        sample_img = f"/workspace/dataset/images/val/{stem}.JPG"
        sample_lbl = f"/workspace/dataset/labels/val/{stem}.txt"
        img = cv2.imread(sample_img)
    
    if img is None:
        print(f"{stem}.JPG — NOT FOUND, skipping")
        continue
    
    h, w = img.shape[:2]
    print(f"{stem}.JPG — loaded {w}x{h}")
    
    n_boxes = 0
    if Path(sample_lbl).exists():
        with open(sample_lbl) as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) != 5:
                    continue
                cls, cx, cy, bw, bh = map(float, parts)
                x1 = int((cx - bw/2) * w)
                y1 = int((cy - bh/2) * h)
                x2 = int((cx + bw/2) * w)
                y2 = int((cy + bh/2) * h)
                cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
                n_boxes += 1
    
    fig, ax = plt.subplots(figsize=(8, 14))
    ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    ax.set_title(f"{stem}.JPG — {n_boxes} boxes drawn")
    ax.axis("off")
    
    out_path = OUTPUT_DIR / f"verify_{stem}.png"
    plt.savefig(out_path, dpi=90, bbox_inches="tight")
    plt.close()
    
    print(f"  Saved: {out_path} ({n_boxes} boxes)")

print(f"\nAll verification images saved to: {OUTPUT_DIR}")
```

```python
from ultralytics import YOLO
import shutil
from pathlib import Path


runs_dir = Path("runs")
if runs_dir.exists():
    shutil.rmtree(runs_dir)
    print("Cleared old runs")

# Train
model = YOLO("yolo11l.pt")

results = model.train(
    data="/workspace/dataset/data.yaml",
    project="runs/detect",
    name="tufts_clean_v1",
    
    # Full training schedule
    epochs=150,
    patience=30,
    
    # H200 can handle larger batch
    batch=32,
    imgsz=1024,
    device=0,
    workers=8,
    
    # Optimizer
    optimizer="AdamW",
    lr0=0.001,
    lrf=0.01,
    cos_lr=True,
    warmup_epochs=3,
    
    # Clinical augmentations (keep rotation for robustness)
    hsv_h=0.0, hsv_s=0.0, hsv_v=0.2,
    degrees=10.0,
    translate=0.1,
    scale=0.2,
    fliplr=0.5,
    flipud=0.0,
    mosaic=0.0,
    mixup=0.0,
    
    # Reproducibility
    deterministic=True,
    seed=42,
    
    # Logging
    plots=True,
    verbose=True,
)

print("\n" + "="*60)
print("TRAINING COMPLETE")
print("="*60)
print(f"Best weights: runs/detect/tufts_clean_v1/weights/best.pt")
```
