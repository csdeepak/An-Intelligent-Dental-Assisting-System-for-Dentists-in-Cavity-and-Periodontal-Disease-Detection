```python
# Cell 4: train
from ultralytics import YOLO

model = YOLO("yolo11l.pt")  # auto-downloads

results = model.train(
    data="/workspace/dataset/data.yaml",
    project="runs/detect",
    name="tufts_v1",
    epochs=150,
    patience=30,
    batch=48, # H200
    imgsz=1024,
    device=0,
    workers=8,
    optimizer="AdamW",
    lr0=0.001,
    lrf=0.01,
    cos_lr=True,
    # Clinical augmentations
    hsv_h=0.0, hsv_s=0.0, hsv_v=0.2,
    degrees=10.0, translate=0.1, scale=0.2,
    shear=0.0, perspective=0.0,
    fliplr=0.5, flipud=0.0,
    mosaic=0.0, mixup=0.0, copy_paste=0.0,
    plots=True,
    verbose=True,
)
```

```python
# Cell 5: evaluate
from ultralytics import YOLO
from pathlib import Path

model = YOLO("runs/detect/tufts_v1/weights/best.pt")

metrics = model.val(
    data="/workspace/dataset/data.yaml",
    split="test",
    project="runs/detect",
    name="tufts_v1_eval",
    imgsz=1024,
    batch=16,
    device=0,
    plots=True,
)

# Print the key metrics
print(f"\nTEST SET METRICS:")
print(f"  Precision:     {metrics.box.mp:.4f}")
print(f"  Recall:        {metrics.box.mr:.4f}")
print(f"  mAP@0.5:       {metrics.box.map50:.4f}")
print(f"  mAP@0.5:0.95:  {metrics.box.map:.4f}")
```
